4a. Plot a KDE of ejection_failure

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("data/heart_failure_clinical_records_dataset.csv")

sns.kdeplot(df["ejection_fraction"], fill=True, bw_adjust=0.5)
plt.title("Kernel Density Estimate of Ejection Fraction")
plt.xlabel("Ejection Fraction (%)")
plt.ylabel("Density")
plt.show()
```

4(b) Resample 15 times with replacement

4(b) Resample 15 times with replacement

```python
for i in range(15):
    sample = df.sample(frac=1.0, replace=True)
    sns.kdeplot(sample["ejection_fraction"], alpha=0.3)
plt.title("Bootstrap KDEs of Ejection Fraction (15 Resamples)")
plt.xlabel("Ejection Fraction (%)")
plt.ylabel("Density")
plt.show()
```
